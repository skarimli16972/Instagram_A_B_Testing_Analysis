```python
# Reload the original data first so the cell can be run multiple times safely, names are long so shortening them
df = pd.read_csv('/AB_Testing_Results_ (1).csv')
df["variant"] = ["Usual Audio", "Trending Audio", "Static", "Reel", "Masscott", "Person"]

# The following are the formulas needed to calculate the rates, percentage is the portion of total multiplied by 100
def percentage (part, total):
    return part / total * 100

# Link CTR if link clicks / by impressions
df["Link_ctr"] = percentage(df["Link clicks"], df["Impressions"])

# Follow rate = new follow divided / profile visits
df["Follow_rate"] = percentage(df["Instagram follows"], df["Instagram profile visits"])

# Keep only the columns needed for comparison
df = df[["variant", "Link_ctr", "Follow_rate"]]
df.round(2)
```

```python
# matplotlib is the library for drawing charts
import matplotlib.pyplot as plt

# bar chart of link CTR for each variant
plt.bar(df["variant"], df["Link_ctr"])
plt.xticks(rotation=90)
plt.ylabel("Link CTR (%)")
plt.show()
```

I wanted to find which type of content works better on Instagram through 3 A/B Tests, I was evaluating trending audio vs. usual, static vs.reel and masscott content vs real person. First I loaded CSV file to pandas and shortened title for direct components. CTR column had different scales on rows so for better precision I use link CTR and Follow rate through formula and compared them .
